d. Use functions to perform deposit, withdrawal, and balance inquiry

```python
def withdraw(bal):
  with_amt=float(input('Enter the amount to withdraw: '))
  if bal>with_amt:
    bal=bal-with_amt
    print(with_amt,' ruppees withdrawed')
  else:
    print('Insufficient balance!!!')
  return bal
```

```python
def deposit(bal):
  dep_amt=float(input('Enter the amount to deposit: '))
  if dep_amt>10000:
    print('This amount exceeds the one time limit')
  else:
    bal=dep_amt+bal
  return bal
```

```python
balance=10000.0
print(' Welcome to ABC bank\n----------------------\nWhat can we do for you today?\n')
print('1. Withdraw cash\n2. Deposit cash\n3. Check Balance\n4. Quit ')
while True:
  option=int(input('Choose your option (1-4): '))
  if option==1:
    balance=withdraw(balance)
  elif option==2:
    balance=deposit(balance)
  elif option==3:
    print('Your current balance is: ',balance)
  elif option==4:
    print('Thank you...')
    break
  else:
    print('Select a valid option')
```

3. Password Validator
Write a function that checks if a password is strong: at least 8 characters,
contains at least one digit, one uppercase letter, and one special character
(!@#$%).

```python
password=input('Enter password')
dig=0
upper=0
symb=0
symbol='!@#$%'
if len(password)>=8:
  for i in password:
    if i.isdigit():
      dig=dig+1
    if i.isupper():
      upper=upper+1
    if i in symbol:
      symb=symb+1
  if dig>0 and upper>0 and symb>0:
    print('Strong Password')
  else:
    print('Weak password!')
else:
  print('weak password!!')


```

5. FizzBuzz Variant

Print numbers 1 to 50, but replace multiples of 3 with "Fizz", multiples of 5
with "Buzz", multiples of both with "FizzBuzz" — and multiples of 7 with
"Bang".

```python
 for i in range(1,51):
  if i%3==0 and i%5==0:
    print('FizzBuzz')
  elif i%3==0:
    print('Fizz')
  elif i%5==0:
    print('Buzz')
  elif i%7==0:
    print('Bang')
  else:
    print(i)
```
